## 2. Download videos

The part will automatically download the videos from Huggingface to `work_dir/data`

### Getting video paths

Change the index of videos you are going to download. The program will download the videos from paths indicated in the `website_files.txt`

> NOTE: Be aware of the number of files to be downloaded. You should be sure that:
> * Your local storage space is large enough to hold this amount of raw video files
> * Your CPU or GPU RAM is large enough to process.
>
> We've tried our best to optimize memory usage. Based on our experiment, 100 videos (from GUI-WORLD) would fit in Colab environment with T4 GPU (~15 GB)

```python
##### Change the idex of video you want to generate #####
file_path = "website_files.txt"
start = 0
end = 50
###################################################################

# Below will download the video from start ~ end - 1
video_list = []
with open(file_path, "r") as f:
  video_list = f.readlines()

print(f"Total number of videos: {len(video_list)}")
print(f"Dowloading videos from index {start} to {end - 1}. Number of downloads: {end - start}")
video_list = video_list[start:end]
video_list = [line.strip() for line in video_list]
```

```python
video_list
```

This will start downloading the videos.

```python
from huggingface_hub import hf_hub_download

data_path = f"data{start}-{end-1}"
for video in video_list:
    hf_hub_download(repo_id="shuaishuaicdp/GUI-World", filename=video, repo_type="dataset", local_dir=data_path)
```

## 3. Generate embeddings

### Swin Transformer

This part will take up to 5 minutes for each 50 videos. You can increase the `batch_size` in `SWin` for faster embedding, but be aware that it might cause GPU overhead.

```python
from video2tensor import *

set_seed()

videos_folder = data_path + '/' + video_list[0].split("/")[0]
ft_folder = f"features/swin{start}-{end-1}"
target_fps = 30
batch_size = (256,256)
embedding = SWin(batch_size=64)
embedding.set_model(model_name="test")

video2tensor(videos_folder=videos_folder,target_fps=target_fps,batch_size=batch_size,ft_folder=ft_folder,embedding=embedding)

# Delete caches
torch.cuda.empty_cache()
```

```python
import shutil

# This will zip the embeddings and delete the files
shutil.make_archive(ft_folder, 'zip', ft_folder)
shutil.rmtree(ft_folder)
```

### I3D

This part will takes about 1 minutes for each 50 videos.


> NOTE: You will need to download the pretrained model from https://github.com/piergiaj/pytorch-i3d/blob/master/models/rgb_imagenet.pt

```python
from video2tensor import *

set_seed()

videos_folder = data_path + '/' + video_list[0].split("/")[0]
ft_folder = f"features/i3d{start}-{end-1}"
target_fps = 30
batch_size = (224, 224)
embedding = I3D()

video2tensor(videos_folder=videos_folder,target_fps=target_fps,batch_size=batch_size,ft_folder=ft_folder,embedding=embedding)

# Delete caches
torch.cuda.empty_cache()
```

```python
import shutil

# This will zip the embeddings and delete the files
shutil.make_archive(ft_folder, 'zip', ft_folder)
shutil.rmtree(ft_folder)
```
